```python
batch_size = 8
train_generator = train_data_gen.flow_from_dataframe(
    dataframe=final_df,
    x_col="Image",
    y_col="Label",
    target_size=(224, 224),
    batch_size=batch_size,
    class_mode="categorical",#sparse
    subset='training',
    shuffle=True,
    seed=42
)
valid_generator = train_data_gen.flow_from_dataframe(
    dataframe=final_df,
    x_col="Image",
    y_col="Label",
    target_size=(224, 224),
    class_mode="categorical", #sparse
    subset='validation',
    shuffle=True,
    seed=42 
)
test_generator = test_data_gen.flow_from_dataframe(
    dataframe=final_test_df,
    x_col="Image",
    y_col="Label",
    target_size=(224, 224),
    batch_size=1,
    class_mode='categorical',
    shuffle=False,
)
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import seaborn as sns
import random
from PIL import Image
from tensorflow.keras.preprocessing import image
from tensorflow.keras.applications.vgg16 import VGG16, preprocess_input
from tensorflow.keras.models import Model
from sklearn.preprocessing import LabelEncoder
from sklearn.utils import resample
from sklearn.svm import SVC
from sklearn.metrics import classification_report, ConfusionMatrixDisplay, confusion_matrix, accuracy_score
```

```python
import os
from tensorflow.keras.preprocessing.image import load_img, img_to_array
import numpy as np

# Path to the training data
data_path = "train"

# Specify target size for resizing images
TARGET_SIZE = (224, 224)  # Example size, adjust as needed

# Function to process images and extract features
def process_images(data_path):
    features = []
    labels = []
    
    for root, dirs, files in os.walk(data_path):
        for file in files:
            try:
                # Full path to the image
                image_path = os.path.join(root, file)
                
                # Load the image
                img = load_img(image_path, target_size=TARGET_SIZE)
                
                # Convert to numpy array
                img_array = img_to_array(img)
                
                # Normalize the pixel values
                img_array /= 255.0
                
                # Add to features list
                features.append(img_array)
                
                # Extract label from folder name
                label = os.path.basename(root)
                labels.append(label)
            except Exception as e:
                print(f"Error processing {image_path}: {e}")
    
    return np.array(features), np.array(labels)

# Extract features and labels
features, labels = process_images(data_path)

# Output the shape of the features and labels for verification
print(f"Features shape: {features.shape}")
print(f"Labels shape: {labels.shape}")
```

```python
# Define class folders
class_folders = [
    'Acne and Rosacea Photos',
    'Actinic Keratosis Basal Cell Carcinoma and other Malignant Lesions',
    'Melanoma Skin Cancer Nevi and Moles'
]

# Define base paths
train_base_path = 'train'
test_base_path = 'test'

# Function to get image paths and labels
def get_image_paths_and_labels(base_path, class_folders):
    image_paths = []
    labels = []
    for folder in class_folders:
        folder_path = os.path.join(base_path, folder)
        for root, _, files in os.walk(folder_path):
            for file in files:
                if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                    image_paths.append(os.path.join(root, file))
                    labels.append(folder)
    return pd.DataFrame({'Image': image_paths, 'Label': labels})

# Load train and test data
train_df = get_image_paths_and_labels(train_base_path, class_folders)
test_df = get_image_paths_and_labels(test_base_path, class_folders)

print(f"Number of training samples: {train_df.shape[0]}")
print(f"Number of testing samples: {test_df.shape[0]}")

```

```python
def plot_sample_images(df, title, nrows=1, ncols=5):
    fig = plt.figure(figsize=(20, 5))
    for i in range(nrows * ncols):
        ax = fig.add_subplot(nrows, ncols, i + 1)
        img_path = df['Image'].iloc[i]
        img = image.load_img(img_path, target_size=(224, 224))
        img = image.img_to_array(img)
        ax.imshow(img / 255.)
        ax.set_title(df['Label'].iloc[i])
        ax.axis('off')
    plt.suptitle(title, fontsize=16)
    plt.show()

# Plot sample images from training set
plot_sample_images(train_df, 'train')
```

```python
# Encode labels as integers
label_encoder = LabelEncoder()
train_df['Label_encoded'] = label_encoder.fit_transform(train_df['Label'])
test_df['Label_encoded'] = label_encoder.transform(test_df['Label'])

# Display class distribution before balancing
plt.figure(figsize=(8,6))
sns.countplot(x='Label', data=train_df)
plt.title('Class Distribution Before Balancing')
plt.show()

# Function to balance dataset by oversampling
def balance_dataset(df):
    max_count = df['Label'].value_counts().max()
    balanced_df = pd.DataFrame()
    for label in df['Label'].unique():
        label_df = df[df['Label'] == label]
        if len(label_df) < max_count:
            label_df = resample(label_df, replace=True, n_samples=max_count, random_state=42)
        balanced_df = pd.concat([balanced_df, label_df], ignore_index=True)
    return balanced_df

# Balance the training data
balanced_train_df = balance_dataset(train_df)

# Display class distribution after balancing
plt.figure(figsize=(8,6))
sns.countplot(x='Label', data=balanced_train_df)
plt.title('Class Distribution After Balancing')
plt.show()
```

```python
# Initialize VGG16 model for feature extraction
vgg_model = VGG16(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
vgg_model.trainable = False  # Freeze the model

# Create a new model that outputs the flattened features
feature_extractor = Model(inputs=vgg_model.input, outputs=tf.keras.layers.GlobalAveragePooling2D()(vgg_model.output))

def extract_features(df, model):
    features = []
    for img_path in df['Image']:
        try:
            img = load_img(img_path, target_size=(224, 224))
            img_array = img_to_array(img)
            img_array = preprocess_input(img_array)
            img_array = np.expand_dims(img_array, axis=0)  # Create batch axis
            feature = model.predict(img_array)
            features.append(feature.flatten())
        except Exception as e:
            print(f"Error processing {img_path}: {e}")
            features.append(np.zeros(model.output_shape[1]))
    return np.array(features)

# Extract features for training and testing data
print("Extracting features from training data...")
X_train = extract_features(balanced_train_df, feature_extractor)
y_train = balanced_train_df['Label_encoded'].values

print("Extracting features from testing data...")
X_test = extract_features(test_df, feature_extractor)
y_test = test_df['Label_encoded'].values

print(f"Feature shape: {X_train.shape}")
```

```python
# Initialize and train the SVM classifier
svm_classifier = SVC(kernel='linear', probability=True, random_state=42)
print("Training SVM classifier...")
svm_classifier.fit(X_train, y_train)
print("Training completed.")
```
